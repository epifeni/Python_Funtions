```python
# function that checks if a nubmer is even
def is_even(num):
    """Returns True if the number is even, False otherwise."""
    return num % 2 == 0
```

```python
is_even(4)
```

```python
is_even(5)
```

```python
# Return true if any number is even in a list
def any_even(numbers):
    """Returns True if any number in the list is even, False otherwise."""
    for num in numbers:
        if is_even(num):
            return True
    return False
```

```python
# Pass in a list with all odd numbers
any_even([1, 3, 5, 7])  # Should return False
```

```python
# Function to return all the even numbers in a list
def all_even(numbers):
    """Returns a list of all even numbers from the input list."""
    even_numbers = []
    for num in numbers:
        if is_even(num):
            even_numbers.append(num)
        else:
            pass
    return even_numbers
```

# Tuple unpacking

```python
work_hours = [("Alice", 100), ("Billy", 400), ("Cassie", 800)]
def employee_check(work_hours):
    """Returns the name of the employee with the highest work hours."""
    current_max = 0
    employee_of_month = ""

    for employee, hours in work_hours:
        if hours > current_max:
            current_max = hours
            employee_of_month = employee
        else:
            pass

    return (employee_of_month, current_max)
```

```python
result = employee_check(work_hours)
```

```python
result
```

```python
# Tuple unpacking example
name, hours = employee_check(work_hours)
```

```python
result
```

```python
name
```

```python
hours
```

```python
name, hours, location = employee_check(work_hours)
```

# Function that can take in a player guess

```python
mylist = [' ','O',' ']
```

```python
shuffle_list(mylist)
```

```python
def player_guess():
    """Function that can take in a player guess and return it as an integer."""
    guess = ''
    while guess not in ['0', '1', '2']:
        guess = input("Pick a number: 0, 1, or 2: ")
    return int(guess)
```

```python
player_guess()
```

```python
myindex = player_guess()
```

```python
myindex
```

```python
def check_guess(mylist, guess):
    """Checks if the player's guess is correct and prints the result."""
    if mylist[guess] == 'O':
        print("Correct!")
    else:
        print("Wrong guess.")
        print(mylist)
```

```python
# Function interaction example

# Initial list
mylist = [' ', 'O', ' ']


# Shuffle the list
mixedup_list = shuffle_list(mylist)

# User guess
guess = player_guess()

# Check guess
check_guess(mixedup_list, guess)
```
